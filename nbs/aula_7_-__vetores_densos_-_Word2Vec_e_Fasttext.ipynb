#Word2Vec

##1. O que é Word2Vec?

Word2Vec (Mikolov et al., 2013) é um grupo de modelos que criam "word embeddings", ou seja, representações vetoriais de palavras. A ideia central é que palavras com significados semelhantes (ou que aparecem em contextos semelhantes) terão vetores semelhantes no espaço vetorial.

Existem duas arquiteturas principais no Word2Vec:

    CBOW (Continuous Bag-of-Words): Prediz a palavra alvo a partir de seu contexto (as palavras ao redor).
    Skip-gram: Prediz as palavras do contexto a partir da palavra alvo.

Ambas as abordagens visam capturar relações semânticas e sintáticas entre as palavras.

##2. Por que usar modelos pré-treinados?

Treinar um modelo Word2Vec robusto do zero exige um corpus textual grande e capacidade computacional, já os modelos pré-treinados já foram treinados em enormes volumes de texto, e seus vetores já capturaram diversas nuances da língua.

Vantagens de usar modelos pré-treinados:

    Economia de tempo e recursos: Não é necessário treinar o modelo.
    Melhor desempenho: Os modelos já aprenderam representações de palavras.

##4. Download e Carregamento dos Modelos do NILC

O NILC disponibiliza diversos modelos Word2Vec pré-treinados para o português, treinados em diferentes corpora (como Wikipedia, Notícias, etc.) e com diferentes tamanhos de vetores (dimensões).

Vamos baixar um modelo comum, por exemplo, o modelo "cbow_s300.zip" (CBOW, 300 dimensões, treinado com notícias e Wikipedia). Você pode encontrar outros modelos e suas descrições no site do NILC: http://www.nilc.icmc.usp.br/nilc/index.php/repositorio-de-modelos-word2vec

```python
#Célula criada por motivo de erro na biblioteca numpy

!pip uninstall -y numpy
!pip install numpy==1.24.4 --no-cache-dir --force-reinstall
```

```python
import wget
import zipfile
import os
from gensim.models import KeyedVectors

model_url = "http://143.107.183.175:22980/download.php?file=embeddings/word2vec/cbow_s100.zip"
model_name = "cbow_s100.zip"
extracted_model_path = "cbow_s100.txt" # O arquivo .txt que gensim lê

# Baixar o modelo
if not os.path.exists(model_name):
    print(f"Baixando o modelo {model_name}...")
    wget.download(model_url)
    print("Download concluído!")
else:
    print(f"O modelo {model_name} já existe.")

# Descompactar o modelo
if not os.path.exists(extracted_model_path):
    print(f"Descompactando {model_name}...")
    with zipfile.ZipFile(model_name, 'r') as zip_ref:
        zip_ref.extractall('.')
    print("Descompactação concluída!")
else:
    print(f"O arquivo {extracted_model_path} já existe.")

# Carregar o modelo Word2Vec
print("Carregando o modelo Word2Vec...")
# O formato de arquivo .txt do NILC é compatível com load_word2vec_format
model = KeyedVectors.load_word2vec_format(extracted_model_path, binary=False)
print("Modelo carregado com sucesso!")
```

##5. Exemplos de Utilização

Agora que o modelo está carregado, podemos explorar suas funcionalidades.

###5.1. Acessando Vetores de Palavras

Cada palavra no vocabulário do modelo tem um vetor associado.

```python
# Obter o vetor de uma palavra
word = "brasil"
try:
    vector = model[word]
    print(f"Vetor da palavra '{word}':")
    print(vector[:10]) # Exibir os primeiros 10 elementos do vetor
    print(f"Dimensão do vetor: {len(vector)}")
except KeyError:
    print(f"A palavra '{word}' não está no vocabulário do modelo.")

word2 = "computador"
try:
    vector2 = model[word2]
    print(f"\nVetor da palavra '{word2}':")
    print(vector2[:10])
except KeyError:
    print(f"A palavra '{word2}' não está no vocabulário do modelo.")
```

###5.2. Encontrando Palavras Semelhantes (Similaridade Cosseno)

Uma das aplicações mais poderosas do Word2Vec é encontrar palavras semanticamente semelhantes. O modelo calcula a similaridade cosseno entre os vetores das palavras. Quanto mais próximo de 1, mais semelhantes são as palavras.

```python
# Encontrar palavras mais semelhantes a uma dada palavra
word = "rei"
print(f"\nPalavras mais semelhantes a '{word}':")
for similar_word, score in model.most_similar(word):
    print(f"  {similar_word}: {score:.4f}")

print("\nPalavras mais semelhantes a 'mulher':")
for similar_word, score in model.most_similar("mulher"):
    print(f"  {similar_word}: {score:.4f}")

print("\nPalavras mais semelhantes a 'cachorro':")
for similar_word, score in model.most_similar("cachorro"):
    print(f"  {similar_word}: {score:.4f}")
```

###5.3. Analogias (Operações Vetoriais)

A característica mais fascinante do Word2Vec é a capacidade de realizar operações aritméticas com vetores de palavras para resolver analogias, como "rei - homem + mulher = rainha".

```python
# Exemplo de analogia: rei - homem + mulher = ?
result = model.most_similar(positive=['mulher', 'rei'], negative=['homem'])
print(f"\nRei - Homem + Mulher = ?")
for word, score in result:
    print(f"  {word}: {score:.4f}")

# Outro exemplo: brasil - brasília + londres = ? (país - capital_país + capital_estrangeira = país_estrangeiro)
result2 = model.most_similar(positive=['londres', 'brasil'], negative=['brasília'])
print(f"\nBrasil - Brasília + Londres = ?")
for word, score in result2:
    print(f"  {word}: {score:.4f}")

# Exemplo: médico - hospital + escola = ? (profissão - local_trabalho + novo_local = nova_profissão_nesse_local)
result3 = model.most_similar(positive=['escola', 'médico'], negative=['hospital'])
print(f"\nMédico - Hospital + Escola = ?")
for word, score in result3:
    print(f"  {word}: {score:.4f}")
```

###5.4. Palavra Fora do Contexto (Odd-one-out)

Podemos identificar a palavra que menos se encaixa em um conjunto de palavras.

```python
# Identificar a palavra que não pertence ao conjunto
words = ["maçã", "banana", "laranja", "carro"]
try:
    odd_one = model.doesnt_match(words)
    print(f"\nPalavra que não pertence ao conjunto {words}: '{odd_one}'")
except Exception as e:
    print(f"\nErro ao encontrar a palavra que não pertence ao conjunto {words}: {e}")

words2 = ["professor", "aluno", "livro", "árvore"]
try:
    odd_one2 = model.doesnt_match(words2)
    print(f"\nPalavra que não pertence ao conjunto {words2}: '{odd_one2}'")
except Exception as e:
    print(f"\nErro ao encontrar a palavra que não pertence ao conjunto {words2}: {e}")
```

###5.5. Similaridade entre duas palavras

Podemos calcular diretamente a similaridade entre duas palavras.

```python
# Calcular a similaridade entre duas palavras
word1 = "amor"
word2 = "ódio"
try:
    similarity = model.similarity(word1, word2)
    print(f"\nSimilaridade entre '{word1}' e '{word2}': {similarity:.4f}")
except KeyError as e:
    print(f"\nErro: {e}. Certifique-se de que ambas as palavras estão no vocabulário.")

word3 = "gato"
word4 = "felino"
try:
    similarity2 = model.similarity(word3, word4)
    print(f"Similaridade entre '{word3}' e '{word4}': {similarity2:.4f}")
except KeyError as e:
    print(f"\nErro: {e}. Certifique-se de que ambas as palavras estão no vocabulário.")
```

##6. Limitações do Word2Vec

Apesar de sua utilidade, o Word2Vec possui algumas limitações:

*    Ambiguidade Semântica (Polissemia): Uma palavra com múltiplos significados (ex: "banco" - instituição financeira vs. assento) terá apenas um vetor.
*    Palavras Fora do Vocabulário (Out-of-Vocabulary - OOV): Se uma palavra não foi vista durante o treinamento, o modelo não terá um vetor para ela.
*    Representação de Frases/Sentenças: O Word2Vec gera embeddings para palavras individuais, não para sentenças completas. Para isso, são necessárias técnicas adicionais.
*    Necessidade de Grande Corpus: Para treinar um modelo do zero, é necessário um corpus textual muito grande.

#Fasttext

##1. O que é FastText?
FastText foi desenvolvido pelo Facebook AI Research (FAIR) em 2016. Enquanto o Word2Vec aprende representações vetoriais para palavras inteiras, o FastText vai além, representando cada palavra como um conjunto de n-gramas de caracteres. Isso permite:

Tratar palavras desconhecidas (OOV - Out-of-Vocabulary): Mesmo que uma palavra não tenha sido vista durante o treinamento, o FastText pode inferir um vetor para ela com base em seus subcomponentes.

Melhor representação para palavras raras: Palavras com poucas ocorrências se beneficiam dos n-gramas compartilhados com outras palavras.

Melhor desempenho para idiomas morfologicamente ricos: Idiomas com muitas flexões (como português) se beneficiam da representação por subunidades.

Assim como o Word2Vec, o FastText possui duas variantes arquiteturais:

    CBOW (Continuous Bag-of-Words)
    Skip-gram

##2. Por que usar modelos pré-treinados?
Assim como no Word2Vec, os modelos FastText pré-treinados oferecem:

    Economia de tempo e recursos: Evita o custo computacional de treinar um modelo do zero
    Melhor desempenho: Os modelos já aprenderam representações de palavras.
    Pronto para uso: Ideal para prototipagem rápida e aplicações onde o treinamento customizado não é viável

##4. Download e Carregamento dos Modelos FastText
Vamos utilizar os modelos em português disponibilizados pelo NILC (http://nilc.icmc.usp.br/nilc/index.php/repositorio-de-word-embeddings-do-nilc). Desta vez, usaremos um modelo FastText.

```python
import wget
import os
from gensim.models import FastText, KeyedVectors
```

```python
# Configurações do modelo
model_url = "http://143.107.183.175:22980/download.php?file=embeddings/fasttext/cbow_s100.zip"
model_name = "skip_s100.zip"
extracted_model_path = "skip_s100.vec"  # Arquivo .vec que o gensim lê

# Baixar o modelo
if not os.path.exists(model_name):
    print(f"Baixando o modelo {model_name}...")
    wget.download(model_url, model_name)
    print("\nDownload concluído!")
else:
    print(f"O modelo {model_name} já existe.")

# Descompactar o modelo
if not os.path.exists(extracted_model_path):
    print(f"Descompactando {model_name}...")
    import zipfile
    with zipfile.ZipFile(model_name, 'r') as zip_ref:
        zip_ref.extractall('.')
    print("Descompactação concluída!")
else:
    print(f"O arquivo {extracted_model_path} já existe.")

# Carregar o modelo FastText
print("Carregando o modelo FastText...")
try:
    # FastText models can be loaded directly with load_facebook_vectors
    model = KeyedVectors.load_word2vec_format(extracted_model_path)
    print("Modelo carregado com sucesso!")
except Exception as e:
    print(f"Erro ao carregar o modelo: {e}")
```

##5. Exemplos de Utilização

###5.1. Acessando Vetores de Palavras

```python
# Obter o vetor de uma palavra
word = "brasil"
try:
    vector = model[word]
    print(f"Vetor da palavra '{word}':")
    print(vector[:10])  # Exibir os primeiros 10 elementos do vetor
    print(f"Dimensão do vetor: {len(vector)}")
except KeyError:
    print(f"A palavra '{word}' não está no vocabulário do modelo.")
```

###5.2. Encontrando Palavras Semelhantes

```python
# Palavras similares
word = "inteligente"
print(f"\nPalavras mais semelhantes a '{word}':")
try:
    for similar_word, score in model.most_similar(word):
        print(f"  {similar_word}: {score:.4f}")
except KeyError:
    print(f"Palavra '{word}' não encontrada no vocabulário.")
```

###5.3. Analogias Semânticas

```python
# Analogia: rei - homem + mulher = ?
try:
    result = model.most_similar(positive=['mulher', 'rei'], negative=['homem'])
    print("\nRei - Homem + Mulher = ?")
    for word, score in result:
        print(f"  {word}: {score:.4f}")
except KeyError as e:
    print(f"\nErro: {e}. Alguma palavra não está no vocabulário.")
```

###5.4. Tratamento de Palavras Desconhecidas (OOV)
Uma das grandes vantagens do FastText é lidar com palavras não vistas durante o treinamento:

```python
# Testando com palavra desconhecida (possivelmente)
unknown_word = "desinstitucionalização"
try:
    vector = model[unknown_word]
    print(f"\nVetor para palavra desconhecida '{unknown_word}':")
    print(vector[:10])

    # Palavras similares à desconhecida
    print(f"\nPalavras similares a '{unknown_word}':")
    for similar_word, score in model.most_similar(unknown_word):
        print(f"  {similar_word}: {score:.4f}")
except KeyError:
    print(f"\nA palavra '{unknown_word}' não pôde ser representada.")
```

###5.5. Similaridade entre Palavras

```python
# Similaridade entre palavras
word_pairs = [
    ("cão", "gato"),
    ("feliz", "alegre"),
    ("computador", "banana")
]

for w1, w2 in word_pairs:
    try:
        sim = model.similarity(w1, w2)
        print(f"Similaridade entre '{w1}' e '{w2}': {sim:.4f}")
    except KeyError as e:
        print(f"Não foi possível calcular similaridade para '{w1}' e '{w2}': {e}")
```

###5.6. Palavra que Não Pertence (Odd One Out)

```python
# Identificar a palavra que não pertence
word_lists = [
    ["carro", "moto", "avião", "banana"],
    ["brasil", "argentina", "paraguai", "europa"]
]

for words in word_lists:
    try:
        odd_one = model.doesnt_match(words)
        print(f"\nNo conjunto {words}, a palavra que não pertence é: '{odd_one}'")
    except Exception as e:
        print(f"\nErro ao processar {words}: {e}")
```

##6. Comparação com Word2Vec: Vantagens do FastText
Vamos demonstrar algumas situações onde o FastText supera o Word2Vec:


###6.1. Flexões Verbais e Palavras Raras

```python
# Testando com flexões verbais não comuns
verb_forms = ["correr", "correu", "corríamos", "correrão"]

print("\nSimilaridades entre formas verbais:")
for i in range(len(verb_forms)):
    for j in range(i+1, len(verb_forms)):
        try:
            sim = model.similarity(verb_forms[i], verb_forms[j])
            print(f"{verb_forms[i]} - {verb_forms[j]}: {sim:.4f}")
        except KeyError:
            print(f"{verb_forms[i]} ou {verb_forms[j]} não encontrada(s)")
```

###6.2. Palavras com Erros de Digitação

```python
# Testando com palavras mal escritas
misspelled = ["cazamento", "intelijente", "ocazão"]

print("\nRepresentação de palavras com erros:")
for word in misspelled:
    try:
        # Verifica se a palavra está no vocabulário
        if word in model:
            print(f"'{word}' está no vocabulário (vetor disponível)")
        else:
            # Mesmo que não esteja, FastText pode gerar um vetor
            vector = model[word]
            similar_words = model.most_similar(word)
            print(f"\nPalavra '{word}' não estava no vocabulário, mas similaridades são:")
            for similar_word, score in similar_words:
                print(f"  {similar_word}: {score:.4f}")
    except Exception as e:
        print(f"Erro com '{word}': {e}")
```

##7. Limitações do FastText
Apesar das vantagens, o FastText tem algumas limitações:

*   Tamanho do Modelo: Os modelos tendem a ser maiores que os do Word2Vec devido ao armazenamento dos n-gramas.

*   Representação de Palavras Muito Curtas: Palavras com menos caracteres que o tamanho mínimo dos n-gramas podem ter representações pobres.

*   Dependência da Estrutura Superficial: A representação baseada em caracteres pode superestimar similaridades entre palavras ortograficamente parecidas mas semanticamente diferentes.
